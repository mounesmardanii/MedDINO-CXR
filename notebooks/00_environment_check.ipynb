```python
from pathlib import Path

NOTEBOOK_DIR = Path.cwd()
PROJECT_ROOT = NOTEBOOK_DIR.parent
VENV_DIR = PROJECT_ROOT / ".venv"

print("Notebook directory:", NOTEBOOK_DIR)
print("Project root:", PROJECT_ROOT)
print("Virtual environment:", VENV_DIR)

print("Project root exists:", PROJECT_ROOT.exists())
print("Virtual environment exists:", VENV_DIR.exists())
```

```python
import torch
import torchvision

print("PyTorch version:", torch.__version__)
print("Torchvision version:", torchvision.__version__)
print("CUDA runtime:", torch.version.cuda)

cuda_available = torch.cuda.is_available()
print("CUDA available:", cuda_available)

if cuda_available:
    gpu_name = torch.cuda.get_device_name(0)
    gpu_memory = torch.cuda.get_device_properties(0).total_memory / (1024**3)

    print("GPU name:", gpu_name)
    print(f"GPU memory: {gpu_memory:.2f} GB")
else:
    print("PyTorch will currently use the CPU.")
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

a = torch.rand((1000, 1000), device=device)
b = torch.rand((1000, 1000), device=device)
result = a @ b

print("Selected device:", device)
print("Tensor device:", result.device)
print("Result shape:", result.shape)
print("Result mean:", result.mean().item())

del a, b, result
torch.cuda.empty_cache()
```
